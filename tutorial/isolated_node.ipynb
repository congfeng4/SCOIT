```python
def check_hypergraph_isolated_nodes(E: np.array):
    from collections import Counter, defaultdict

    node_in_edges = defaultdict(Counter)
    for i in range(E.shape[0]): # All edges.
        for j in range(E.shape[1]): # Each node in an edge.
            node_in_edges[j][E[i, j]] += 1 # One edge mentions this node.

    # If no edge or just one edge mentions a node, then the node is isolated (no neighbours).
    ok = True
    for node_type, edges in node_in_edges.items():
        isolated_nodes0 = [nid for nid, edge_count in edges.items() if edge_count == 0]
        if isolated_nodes0:
            print('Node type', node_type, 'isolated_nodes0', isolated_nodes0)
            ok = False

        isolated_nodes1 = [nid for nid, edge_count in edges.items() if edge_count == 1]
        if isolated_nodes1:
            print('Node type', node_type, 'isolated_nodes1', isolated_nodes1)
            ok = False

    return ok
```
