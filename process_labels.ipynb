PEA-STA

```python
def load_labels():
    cell_stage = np.array(pd.read_csv("data/PEA_STA/cell_stage.csv", header=None))[0]
    labels = []
    for each in cell_stage:
        day = each.split("_")[1]
        treat = each.split("_")[2]
        if day == "0h":
            labels.append("0h untreated")
        elif day == "6d":
            if treat == "contol":
                labels.append("6d untreated")
            elif treat == "BMP4":
                labels.append("BMP4 treated")

    return labels

json.dump(load_labels(), open('data/PEA_STA/cell_stage.json', 'w'))
```

SCoPE2

```python
def load_data():
    cell_stage = np.array(pd.read_csv("./data/SCoPE2/cell_stage.csv", header=None))[0]
    labels = []
    for each in cell_stage:
        if each == "sc_m0":
            labels.append("Macrophage")
        elif each == "sc_u":
            labels.append("Monocyte")

    return labels

labels = load_data()
json.dump(labels, open('data/SCoPE2/cell_stage.json', 'w'))
```
