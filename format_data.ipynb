## Convert all graphs to Hyper_SAGNN format.

```python
try:
   import sys
   del sys.modules['hgnn.hypergraph']
except:
    pass

from hgnn.hypergraph import *

TRAIN_SIZE = 0.8
```

```python
hg = load_graph('./data/sc_GEM')

convert_to_hypersagnn_format(hg, train_size=TRAIN_SIZE, save_dir='./data/sc_GEM')
```

```python
for subdir in Path('./data').iterdir():
    if not subdir.is_dir():
        continue
    try:
        hg = load_graph(subdir)
    except Exception as e:
        print('Failed to load', subdir, 'except', e)
        continue
    convert_to_hypersagnn_format(hg, TRAIN_SIZE, subdir)
```
